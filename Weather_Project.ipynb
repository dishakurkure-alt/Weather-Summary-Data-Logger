# Weather Summary & Data Logger

## 1. Introduction

```python
print('Weather Summary & Data Logger')
```

## 2. Import Libraries

```python
import requests
from bs4 import BeautifulSoup
import re
from datetime import datetime
```

## 3. Web Scraping  by Using BeautifulSoup

```python
url = "https://www.timeanddate.com/weather/india/pune"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

temp_tag = soup.find("div", class_="h2")
temperature = temp_tag.text.strip() if temp_tag else "N/A"
print("Temperature in Pune:", temperature)
```

## 4. Extract Weather Condition Using RegEx

```python
condition_tag = soup.find("p")
if condition_tag:
    match = re.search(r"[A-Za-z ]+", condition_tag.text)
    condition = match.group().strip() if match else "N/A"
else:
    condition = "N/A"

print("Weather Condition:", condition)
```

## 5. Data Structures

```python
weather_list = [temperature, condition]  #lists

weather_dict = {
    "temperature": temperature,
    "condition": condition,
    "time": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
}

location_info = ("Pune", "India")  #Tuple
weather_set = set(weather_list)    #Set

print(weather_dict)    #Strings
```

## 6. Functions

```python
def classify_temperature(temp_string):
    try:
        temp_value = int(re.findall(r"\d+", temp_string)[0])
        if temp_value > 30:
            return "Hot"
        elif temp_value > 20:
            return "Warm"
        else:
            return "Cold"
    except:
        return "Unknown"

print("Temperature Category:", classify_temperature(temperature))
```

## 7. OOP

```python
class WeatherReport:
    def __init__(self, temp, cond, location):
        self.temp = temp
        self.cond = cond
        self.location = location

    def show(self):
        print(f"Location: {self.location}")
        print(f"Temperature: {self.temp}")
        print(f"Condition: {self.cond}")

report = WeatherReport(temperature, condition, location_info)
report.show()
```

## 8. File Handling

```python
with open("weather_log.txt", "a") as file:
    file.write(str(weather_dict) + "\n")
print("Weather data saved to file.")
```

## 9. Exception Handling

```python
try:
    with open("weather_log.txt", "r") as file:
        content = file.read()
        print("\n=== File Content ===")
        print(content)
except FileNotFoundError:
    print("File not found!")
```
